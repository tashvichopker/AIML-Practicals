```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/minsuk-heo/kaggle-titanic/master/input/train.csv")

df
```

```python
print(df.head())
```

```python
print("Rows and Columns:", df.shape) #display dimensions
```

```python
print(df.columns.tolist()) #display column names
```

```python
print(df.dtypes) #datatypes
```

```python
df.info()
```

```python
print(df.describe()) #statistical summary
```

```python
print(df.isnull().sum()) #missing values
```

```python
print("Duplicates:", df.duplicated().sum()) #duplicates
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) #handle missing values in age
```

```python
df['Embarked'] = df['Embarked'].fillna(
    df['Embarked'].mode()[0]
)
```

```python
df.drop('Cabin', axis=1, inplace=True) #remove cabin column
```

```python
print(df['Age'].head())
```

```python
print(df[['Name', 'Age', 'Sex', 'Fare']].head())
```

```python
old_passengers = df[df['Age'] > 50]

print(old_passengers.head(20))
```

```python
female_passengers = df[df['Sex'] == 'female']

print(female_passengers.head(10))
```

```python
female_survivors = df[
    (df['Sex'] == 'female') & 
    (df['Survived'] == 1)
]

print(female_survivors.head(10))
```

```python
class_1_2 = df[df['Pclass'].isin([1, 2])]

print(class_1_2.head(10))
```

```python
print(df['Sex'].value_counts()) #count by gender
```

```python
print(df.groupby('Sex')['Age'].mean()) #avg age by gender
```

```python
print(df.groupby('Pclass')['Fare'].mean())
```

```python
print(df.groupby('Sex')['Survived'].mean())
```

```python
print(df.groupby('Pclass').agg({
    'Age': 'mean',
    'Fare': 'mean',
    'Survived': 'mean'
}))
```
